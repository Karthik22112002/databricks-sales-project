```python
from pyspark.sql.functions import *

data = [
    (1, "Laptop", 2, 50000),
    (2, "Mobile", 3, 20000),
    (3, "Watch", 2, 10000),
    (4, "Tablet", 1, 20000),
    (5, "Laptop", 2, 50000),
]
schema = ["id", "product", "quantity", "amount"]
df_b = spark.createDataFrame(data, schema)
result = df_b.withColumn("Tax", col("amount") * 0.38)
display(result)

```
